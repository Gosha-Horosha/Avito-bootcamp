## Дубликаты

Сначала дубли пар «текст запроса — объявление».

```python
print("Дублирующихся пар (search_query, item_id):",
      train.duplicated(subset=['search_query', 'item_id']).sum())
```

Часть из них — один и тот же текст в разных городах или с разными фильтрами, то есть разные запросы. Настоящий дубль — совпадение всех полей `search_*` и `item_id`.

```python
search_cols = ['search_query', 'search_location_id', 'search_is_delivery_search',
               'search_infm_params_text', 'search_category']

print("Уникальных комбинаций search_* полей:", train[search_cols].drop_duplicates().shape[0])

dup_mask = train.duplicated(subset=['search_query', 'item_id'], keep=False)
dup_rows = train[dup_mask].sort_values('search_query')

check = dup_rows.groupby(['search_query', 'item_id'])[search_cols[1:]].nunique()
print(check.describe())
print(check[(check > 1).any(axis=1)].head(10))

full_dup = train.duplicated(subset=search_cols + ['item_id'], keep=False)
print("Дублей по ВСЕМ search_* полям + item_id:", full_dup.sum())
```

Полные дубли не идут подряд и стоят на разных местах в файле — похоже на одинаковые клики разных пользователей, а не на технический артефакт. Для валидации их можно удалить.

```python
search_cols = ['search_query', 'search_location_id', 'search_is_delivery_search',
               'search_infm_params_text', 'search_category']
full_key = search_cols + ['item_id']

full_dup_mask = train.duplicated(subset=full_key, keep=False)
full_dups = train[full_dup_mask].sort_values(full_key)

print("Всего строк в полных дублях (все вхождения):", len(full_dups))
print("Из них уникальных комбинаций (сколько 'групп' дублей):",
      full_dups.drop_duplicates(subset=full_key).shape[0])

dup_counts = full_dups.groupby(full_key).size().sort_values(ascending=False)
print(dup_counts.describe())
print(dup_counts.head(10))

top_dup_key = dup_counts.index[0]
example = train[
    (train['search_query'] == top_dup_key[0]) &
    (train['search_location_id'] == top_dup_key[1]) &
    (train['search_is_delivery_search'] == top_dup_key[2]) &
    (train['search_infm_params_text'] == top_dup_key[3]) &
    (train['search_category'] == top_dup_key[4]) &
    (train['item_id'] == top_dup_key[5])
]
print(example)

pairs_count2 = dup_counts[dup_counts == 2].index[:5]
for key in pairs_count2:
    subset = train[
        (train['search_query'] == key[0]) &
        (train['search_location_id'] == key[1]) &
        (train['search_is_delivery_search'] == key[2]) &
        (train['search_infm_params_text'] == key[3]) &
        (train['search_category'] == key[4]) &
        (train['item_id'] == key[5])
    ]
    print(subset[['search_query', 'item_id'] + [c for c in subset.columns if c not in full_key]])
    print('---')
```
